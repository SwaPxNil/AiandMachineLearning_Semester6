Problem - 4: Matrix Operations:
â€¢ For the following arrays:
A = np.array([[3,4],[7,8]]) and B = np.array([[5,3],[2,1]]);
Prove following with Numpy:
1. Prove A.Aâˆ’1 = I.
2. Prove AB Ì¸= BA.
3. Prove (AB)T = BTAT.

```python
import numpy as np
A = np.array([[3,4],[7,8]])
B = np.array([[5,3],[2,1]])

# Prove A.Aâˆ’1 = I.

A_inverse_v2 = np.linalg.inv(A)
mul_A_Ainv_v2 = np.dot(A, A_inverse_v2)
print("Multiplication of A and A_inverse_v2: \n", mul_A_Ainv_v2)
idenity_matrix_v2 = np.eye(A.shape[0])
print("Identity matrix: \n", idenity_matrix_v2)
is_identity_v2 = np.allclose(mul_A_Ainv_v2, idenity_matrix_v2)
if(is_identity_v2):
    print("The matrix multiplication of A and its inverse is an Identity matrix or close to it.")
    print("Hence Proved.")
else:
    print("Not Proved.")

# Prove AB Ì¸= BA.

mul_AB_v2 = np.matmul(A,B)
mul_BA_v2 = np.matmul(B,A)
print("Multiplication of A and B: \n", mul_AB_v2)
print("Multiplication of B and A: \n", mul_BA_v2)
equal_mul_v2 = np.array_equal(mul_AB_v2, mul_BA_v2)
if(not equal_mul_v2):
    print("The matrix multiplication of A and B is not equal to the matrix multiplication of B and A.")
    print("Hence Proved.")
else:
    print("The matrix multiplication of A and B is equal to the matrix multiplication of b and A.")
    print("Not proved.")

# Prove (AB)T = BTAT

A_transpose_v2 = np.transpose(A)
B_transpose_v2 = np.transpose(B)
AB_transpose_v2 = np.transpose(mul_AB_v2)
print("The Transpose of AB ((AB)T) is: \n", AB_transpose_v2)
B_trans_A_trans_v2 = np.matmul(B_transpose_v2, A_transpose_v2)
print("The Transpose of B multiplied with Transpose of A (BTAT) is: \n", B_trans_A_trans_v2)
equal_trans_v2 = np.array_equal(AB_transpose_v2, B_trans_A_trans_v2)
if(equal_trans_v2):
    print("The transpose of AB and transpose of B multiplied with transpose of A is equal.")
    print("Hence Proved.")
else:
    print("Not Proved.")
```

Experiment: How Fast is Numpy?

1. Element-wise Addition
2. Element-wise Multiplication
3. Dot Product
4. Matrix Multiplication

```python
import numpy as np
import random
from timeit import timeit

# Creating two different lists of size_v2 1000000

size_v2 = 1000000
A_list_v2 = random.sample(range(size_v2 + 1), size_v2)
B_list_v2 = random.sample(range(size_v2 + 1), size_v2)
A_array_v2 = np.array(A_list_v2)
B_array_v2 = np.array(B_list_v2)

# Python addition

def python_add(A, B):
    result_v2 = []
    for i in range(len(A)):
        result_v2.append(A[i] + B[i])
    return result_v2
python_time_add_v2 = timeit("python_add(A_list_v2, B_list_v2)", globals=globals(), number=100)
print("Python time for add:", python_time_add_v2)

# Numpy addition

def numpy_add(A, B):
    return A + B
numpy_time_add_v2 = timeit("numpy_add(A_array_v2, B_array_v2)", globals=globals(), number=100)
print("Numpy time for add:", numpy_time_add_v2)

# Python multiplication

def python_mul(A, B):
    result_v2 = []
    for i in range(len(A)):
        result_v2.append(A[i] * B[i])
    return result_v2
python_time_mul_v2 = timeit("python_mul(A_list_v2, B_list_v2)", globals=globals(), number=100)
print("Python time for multiplication:", python_time_mul_v2)

# Numpy multiplication

def numpy_mul(A, B):
    return np.multiply(A, B)
numpy_time_mul_v2 = timeit("numpy_mul(A_array_v2, B_array_v2)",globals=globals(), number=100)
print("Numpy time for multiplication:", numpy_time_mul_v2)

# Python dot product

def python_dot(A, B):
    return sum(a * b for a, b in zip(A, B))
python_time_dot_v2 = timeit("python_dot(A_list_v2, B_list_v2)", globals=globals(), number=100)
print("Python time for dot product:", python_time_dot_v2)

# Numpy dot

def numpy_dot(A, B):
    return np.dot(A, B)
numpy_time_dot_v2 = timeit("numpy_dot(A_array_v2, B_array_v2)", globals=globals(), number=100)
print("Numpy time for dot product:", numpy_time_dot_v2)

# Matrix creation

matrix_size_v2 = 300
Z1_matrix_list_v2 = [A_list_v2[i:i + matrix_size_v2] for i in range(0, matrix_size_v2**2, matrix_size_v2)]
Z2_matrix_list_v2 = [B_list_v2[i:i + matrix_size_v2] for i in range(0, matrix_size_v2**2, matrix_size_v2)]
Z1_matrix_arr_v2 = np.array(Z1_matrix_list_v2)
Z2_matrix_arr_v2 = np.array(Z2_matrix_list_v2)

# Python matrix multiplication

def python_matmul(A, B):
    return [
        [sum(a * b for a, b in zip(row_A, col_B)) for col_B in zip(*B)]
        for row_A in A
    ]
python_time_matmul_v2 = timeit(
    "python_matmul(Z1_matrix_list_v2, Z2_matrix_list_v2)",
    globals=globals(),
    number=10
)
print("Python time for matrix multiplication:", python_time_matmul_v2)

# Numpy matrix multiplication

def numpy_matmul(A, B):
    return np.matmul(A, B)
numpy_time_matmul_v2 = timeit(
    "numpy_matmul(Z1_matrix_arr_v2, Z2_matrix_arr_v2)",
    globals=globals(),
    number=10
)
print("Numpy time for matrix multiplication:", numpy_time_matmul_v2)
```
