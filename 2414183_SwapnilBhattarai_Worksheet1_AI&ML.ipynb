# Exercise - 3: Image Compression and Decompression using PCA.
In this exercise, build a PCA from scratch using explained variance method for image compression task.

You are expected to compute the necessary matrices from the scratch. Dataset: Use image of your
choice.
## 1. Load and Prepare Data:
â€¢ Fetch an image of you choice.{If colour convert to grayscale}

â€¢ Center the dataset - Standaridze the Data.

â€¢ Calculate the covaraince matrix of the Standaridze data.
## 2. Eigen Decomposition and Identifying Principal Components:
â€¢ Compute Eigen Values and Eigen Vectors.

â€¢ Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to
the highest eigenvalues.

â€¢ Identify the Principal Components with the help of cumulative Sum plot.
## 3. Reconstruction and Experiment:

â€¢ **Reconstruction:** Transform the original data by multiplying it with the selected eigenvec-
tors(PCs) to obtain a lower-dimensional representation.

â€¢ **Experiments:** Pick Four different combination of principal components with various ex-
plained variance value and compare the result.

â€¢ Display the Results and Evaluate.

```python
# loading and preparing the data

task3_image = Image.open("/content/drive/MyDrive/Sem6/Lenna.png").convert("L") # converting the loaded image to grayscale
task3_image_array = np.array(task3_image) # creating an image array

print("Shape of the image array :", task3_image_array.shape) # the shape of the original image
```

```python
# Center the dataset - Standaridze the Data.
image_mean = np.mean(task3_image_array, axis=0)
centered_data = task3_image_array - image_mean
```

```python
# compute the covaraince matrix of the Standaridze data.
covariance_matrix = np.cov(centered_data, rowvar=False)
print("The shape of the covariance matrix is: ", covariance_matrix.shape)
```

```python
# Compute Eigen Values and Eigen Vectors.
eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)

print("The shape of the eigenvalues is: ", eigenvalues.shape)
print("The shape of the eigenvectors is: ", eigenvectors.shape)
```

```python
# Sort the eigenvalues in descending order and choose the top k eigenvectors corresponding to the highest eigenvalues
sorted_indices = np.argsort( eigenvalues )[:: -1]
eigenvalues = eigenvalues[sorted_indices]
eigenvectors = eigenvectors[: , sorted_indices]
```

```python
# plotting the variance ratio to understand it
explained_variance_ratio = eigenvalues / np.sum(eigenvalues)

plt.plot(np.cumsum(explained_variance_ratio), color="red")
plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.grid(True)
plt.show()
```

```python
# creating a function to reconstruct image with different k value
def reconstruct_image(X_centered, eigenvectors, k, mean):

    # Select top k eigenvectors
    components = eigenvectors[:, :k]

    # Project data
    compressed_data = np.dot(X_centered, components)
    # Reconstruct
    X_reconstructed = np.dot(compressed_data, components.T) + mean

    return X_reconstructed
```

```python
# with different k values to reconstruct the image with the reconstruct_image function
k_values = [5, 20, 50, 100]

plt.figure(figsize=(12,8))

for i, k in enumerate(k_values):

    reconstructed = reconstruct_image(centered_data, eigenvectors, k, image_mean)

    plt.subplot(2,2,i+1)
    plt.imshow(reconstructed.real, cmap='gray')
    plt.title(f"k = {k}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
